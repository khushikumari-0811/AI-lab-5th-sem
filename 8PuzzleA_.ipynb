```python
import heapq

class Node:
    def __init__(self, board, parent, g, h):
        self.board = board
        self.parent = parent
        self.g = g  # Cost from start to current state
        self.h = h  # Heuristic cost to goal
        self.f = g + h  # Total cost

    def __lt__(self, other):
        return self.f < other.f

def calculate_manhattan_distance(board, goal_board):
    distance = 0
    for i in range(3):
        for j in range(3):
            val = board[i][j]
            if val != 0:
                for gi in range(3):
                    for gj in range(3):
                        if goal_board[gi][gj] == val:
                            distance += abs(i - gi) + abs(j - gj)
    return distance

def get_neighbors(node, goal_board):
    neighbors = []
    board = node.board

    x, y = 0, 0
    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                x, y = i, j
                break

    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dx, dy in directions:
        nx, ny = x + dx, y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_board = [row[:] for row in board]
            new_board[x][y], new_board[nx][ny] = new_board[nx][ny], new_board[x][y]

            h = calculate_manhattan_distance(new_board, goal_board)
            neighbors.append(Node(new_board, node, node.g + 1, h))

    return neighbors

def get_inv_count(arr):
    inv_count = 0
    for i in range(8):
        for j in range(i + 1, 9):
            if arr[i] > 0 and arr[j] > 0 and arr[i] > arr[j]:
                inv_count += 1
    return inv_count

def is_solvable(initial, goal):
    # Flatten boards to calculate inversions relative to the goal layout
    initial_flat = [num for row in initial for num in row]
    goal_flat = [num for row in goal for num in row]

    # Map numbers to their target indices in the goal state
    goal_position_map = {num: i for i, num in enumerate(goal_flat)}

    # Rewrite initial layout based on goal position indices (ignoring 0)
    initial_mapped = [goal_position_map[num] for num in initial_flat if num != 0]

    inv_count = 0
    for i in range(len(initial_mapped)):
        for j in range(i + 1, len(initial_mapped)):
            if initial_mapped[i] > initial_mapped[j]:
                inv_count += 1

    return inv_count % 2 == 0

def solve(initial_board, goal_board):
    if not is_solvable(initial_board, goal_board):
        return None

    open_set = []
    closed_set = set()

    h_init = calculate_manhattan_distance(initial_board, goal_board)
    start_node = Node(initial_board, None, 0, h_init)

    heapq.heappush(open_set, start_node)

    while open_set:
        current = heapq.heappop(open_set)
        board_tuple = tuple(tuple(row) for row in current.board)

        if current.h == 0:
            path = []
            while current:
                path.append(current.board)
                current = current.parent
            return path[::-1]

        closed_set.add(board_tuple)

        for neighbor in get_neighbors(current, goal_board):
            neighbor_tuple = tuple(tuple(row) for row in neighbor.board)
            if neighbor_tuple in closed_set:
                continue
            heapq.heappush(open_set, neighbor)

    return None

def input_board(board_name):
    print(f"\nEnter the {board_name} board configuration (use 0 for blank space):")
    print("Enter 3 space-separated numbers per row (e.g., 1 2 3):")
    board = []
    all_numbers = []

    for i in range(3):
        while True:
            try:
                row_input = input(f"Row {i+1}: ").strip().split()
                if len(row_input) != 3:
                    raise ValueError("Each row must contain exactly 3 numbers.")
                row = [int(num) for num in row_input]

                for num in row:
                    if num < 0 or num > 8:
                        raise ValueError("Numbers must be between 0 and 8.")
                    if num in all_numbers:
                        raise ValueError(f"Duplicate entry found for '{num}'.")
                    all_numbers.append(num)

                board.append(row)
                break
            except ValueError as e:
                print(f"Invalid input: {e}. Please retry.")

    if len(all_numbers) != 9:
        print("Error: The entire board configuration must contain all digits from 0 to 8.")
        return input_board(board_name)

    return board

# Main Execution Loop
if __name__ == "__main__":
    print("--- 8-Puzzle A* Solver ---")
    initial = input_board("INITIAL")
    goal = input_board("GOAL")

    print("\nCalculating path...")
    solution_path = solve(initial, goal)

    if solution_path:
        print(f"\nSolution found in {len(solution_path) - 1} moves:\n")
        for step_num, board in enumerate(solution_path):
            print(f"Step {step_num}:")
            for row in board:
                print(" ".join(map(str, row)))
            print()
    else:
        print("\nThis configuration is mathematically unsolvable from the given initial state to the goal state.")
import heapq

class Node:
    def __init__(self, board, parent, g, h):
        self.board = board
        self.parent = parent
        self.g = g  # Cost from start to current state
        self.h = h  # Heuristic cost to goal
        self.f = g + h  # Total cost

    def __lt__(self, other):
        return self.f < other.f

def calculate_manhattan_distance(board, goal_board):
    distance = 0
    for i in range(3):
        for j in range(3):
            val = board[i][j]
            if val != 0:
                for gi in range(3):
                    for gj in range(3):
                        if goal_board[gi][gj] == val:
                            distance += abs(i - gi) + abs(j - gj)
    return distance

def get_neighbors(node, goal_board):
    neighbors = []
    board = node.board

    x, y = 0, 0
    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                x, y = i, j
                break

    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dx, dy in directions:
        nx, ny = x + dx, y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_board = [row[:] for row in board]
            new_board[x][y], new_board[nx][ny] = new_board[nx][ny], new_board[x][y]

            h = calculate_manhattan_distance(new_board, goal_board)
            neighbors.append(Node(new_board, node, node.g + 1, h))

    return neighbors

def get_inv_count(arr):
    inv_count = 0
    for i in range(8):
        for j in range(i + 1, 9):
            if arr[i] > 0 and arr[j] > 0 and arr[i] > arr[j]:
                inv_count += 1
    return inv_count

def is_solvable(initial, goal):
    # Flatten boards to calculate inversions relative to the goal layout
    initial_flat = [num for row in initial for num in row]
    goal_flat = [num for row in goal for num in row]

    # Map numbers to their target indices in the goal state
    goal_position_map = {num: i for i, num in enumerate(goal_flat)}

    # Rewrite initial layout based on goal position indices (ignoring 0)
    initial_mapped = [goal_position_map[num] for num in initial_flat if num != 0]

    inv_count = 0
    for i in range(len(initial_mapped)):
        for j in range(i + 1, len(initial_mapped)):
            if initial_mapped[i] > initial_mapped[j]:
                inv_count += 1

    return inv_count % 2 == 0

def solve(initial_board, goal_board):
    if not is_solvable(initial_board, goal_board):
        return None

    open_set = []
    closed_set = set()

    h_init = calculate_manhattan_distance(initial_board, goal_board)
    start_node = Node(initial_board, None, 0, h_init)

    heapq.heappush(open_set, start_node)

    while open_set:
        current = heapq.heappop(open_set)
        board_tuple = tuple(tuple(row) for row in current.board)

        if current.h == 0:
            path = []
            while current:
                path.append(current.board)
                current = current.parent
            return path[::-1]

        closed_set.add(board_tuple)

        for neighbor in get_neighbors(current, goal_board):
            neighbor_tuple = tuple(tuple(row) for row in neighbor.board)
            if neighbor_tuple in closed_set:
                continue
            heapq.heappush(open_set, neighbor)

    return None

def input_board(board_name):
    print(f"\nEnter the {board_name} board configuration (use 0 for blank space):")
    print("Enter 3 space-separated numbers per row (e.g., 1 2 3):")
    board = []
    all_numbers = []

    for i in range(3):
        while True:
            try:
                row_input = input(f"Row {i+1}: ").strip().split()
                if len(row_input) != 3:
                    raise ValueError("Each row must contain exactly 3 numbers.")
                row = [int(num) for num in row_input]

                for num in row:
                    if num < 0 or num > 8:
                        raise ValueError("Numbers must be between 0 and 8.")
                    if num in all_numbers:
                        raise ValueError(f"Duplicate entry found for '{num}'.")
                    all_numbers.append(num)

                board.append(row)
                break
            except ValueError as e:
                print(f"Invalid input: {e}. Please retry.")

    if len(all_numbers) != 9:
        print("Error: The entire board configuration must contain all digits from 0 to 8.")
        return input_board(board_name)

    return board

# Main Execution Loop
if __name__ == "__main__":
    print("--- 8-Puzzle A* Solver ---")
    initial = input_board("INITIAL")
    goal = input_board("GOAL")

    print("\nCalculating path...")
    solution_path = solve(initial, goal)

    if solution_path:
        print(f"\nSolution found in {len(solution_path) - 1} moves:\n")
        for step_num, board in enumerate(solution_path):
            print(f"Step {step_num}:")
            for row in board:
                print(" ".join(map(str, row)))
            print()
    else:
        print("\nThis configuration is mathematically unsolvable from the given initial state to the goal state.")
```
